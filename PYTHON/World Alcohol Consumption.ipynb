## 11. Bar plot with error bars

Filter all records where average beverage consumption per person is between **0.5 and 2.50**.

Create a **bar plot with error bars**.

**Your code:**

````python
filtered_df = df[
    (df['Display Value'] >= 0.5) &
    (df['Display Value'] <= 2.50)
]

bar_data = filtered_df.groupby('Beverage Types')['Display Value'].agg(
    ['mean', 'std']
).reset_index()

plt.figure(figsize=(10, 6))

plt.bar(
    bar_data['Beverage Types'],
    bar_data['mean'],
    yerr=bar_data['std'],
    capsize=5
)
plt.xlabel('Beverage Types')
plt.ylabel('Average Consumption per Person')
plt.title('Average Beverage Consumption with Error Bars')
plt.show()
````
